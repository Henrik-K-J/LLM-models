```python
from huggingface_hub import hf_hub_download
import os
from pathlib import Path
import shutil

# Hugging Face -token
hf_token = os.environ.get("HF_TOKEN")

if not hf_token:
    try:
        from google.colab import userdata
        hf_token = userdata.get("HF_TOKEN")
    except Exception:
        hf_token = None

base_models_dir = Path("/content/ComfyUI/models")

# Määritetään kaksi YuE2-mallitiedostoa oikeilla kohdekansioilla
models = [
    {
        "repo_id": "Comfy-Org/YuE2",
        "filename": "checkpoints/yue2_3b_int8_convrot.safetensors",
        "folder": "checkpoints"
    },
        {
        "repo_id": "Comfy-Org/YuE2",
        "filename": "checkpoints/yue2_3b_bf16.safetensors",
        "folder": "checkpoints"
    },
    {
        "repo_id": "Comfy-Org/YuE2",
        "filename": "audio_encoders/sheetsage2_bf16.safetensors",
        "folder": "audio_encoders"
    }
]

try:
    for item in models:
        repo_id = item["repo_id"]
        filename = item["filename"]
        folder = item["folder"]

        dest_dir = base_models_dir / folder
        dest_dir.mkdir(parents=True, exist_ok=True)
        final_path = dest_dir / Path(filename).name

        # Jos tiedosto on jo olemassa, hypätään yli
        if final_path.exists():
            if final_path.stat().st_size > 1000000:
                print(f"  Tiedosto on jo olemassa: {final_path}")
                continue
            else:
                final_path.unlink()

        print(f"\nℹ  Ladataan: {filename} -> ComfyUI/models/{folder}/")

        # Ladataan tiedosto lokaalisti ilman symlinkkejä
        downloaded = hf_hub_download(
            repo_id=repo_id,
            filename=filename,
            repo_type="model",
            token=hf_token,
            local_dir=str(dest_dir),
            local_dir_use_symlinks=False
        )

        # Varmistetaan oikea sijoituspaikka
        downloaded_path = Path(downloaded)
        if downloaded_path.exists() and downloaded_path != final_path:
            if final_path.exists():
                final_path.unlink()
            shutil.move(str(downloaded_path), str(final_path))

        if final_path.exists():
            print(f"✅ Tallennettu: {final_path} ({final_path.stat().st_size / (1024**3):.2f} GB)")

    print("\n✅ YuE2-mallit ladattu onnistuneesti oikeisiin kansioihin!")

except Exception as e:
    print(f"\n❌ Virhe latauksessa: {e}")
```

```python
import os
from pathlib import Path

output_path = Path("/content/ComfyUI/output")

if not output_path.exists():
    print(f"❌ Kansiota {output_path} ei löydy.")
else:
    # Suodatetaan pois piilotiedostot ja vakiopidike (.gitkeep tms)
    files = [f for f in os.listdir(output_path) if not f.startswith('.') and f != "_output_images_will_be_put_here"]
    if len(files) == 0:
        print("✅ Output-kansio on tyhjä!")
    else:
        print(f"📂 Kansio ei ole tyhjä. Löytyi {len(files)} tiedostoa:")
        for f in files:
            print(f"  - {f}")
```

```python
import os
from pathlib import Path

audio_output_path = Path("/content/ComfyUI/output/audio")

if not audio_output_path.exists():
    print(f"❌ Kansiota {audio_output_path} ei löydy.")
elif not audio_output_path.is_dir():
    print(f"⚠️ Kohde {audio_output_path} on olemassa, mutta se ei ole kansio.")
else:
    files = os.listdir(audio_output_path)
    if len(files) == 0:
        print("✅ Kansio 'output/audio' on tyhjä!")
    else:
        print(f"📂 Kansio 'output/audio' ei ole tyhjä. Löytyi {len(files)} tiedostoa:")
        for f in sorted(files):
            file_path = audio_output_path / f
            if file_path.is_file():
                size_mb = file_path.stat().st_size / (1024 * 1024)
                print(f"  📄 {f} ({size_mb:.2f} MB)")
            else:
                print(f"  📁 {f}/")
```
