```python
from huggingface_hub import hf_hub_download
import os
from pathlib import Path
import shutil

# Hugging Face -token
hf_token = os.environ.get("HF_TOKEN")

if not hf_token:
    try:
        from google.colab import userdata
        hf_token = userdata.get("HF_TOKEN")
    except Exception:
        hf_token = None

base_models_dir = Path("/content/ComfyUI/models")

# Määritetään tiedostot ja sijoitetaan ne ComfyUI:n standardeihin vakiokansioihin
models = [
    {
        "repo_id": "Lightricks/LTX-2.5",
        "filename": "vae/ltx-2.5-audio-vae-bf16.safetensors",
        "folder": "vae"
    },
    {
        "repo_id": "Lightricks/LTX-2.5",
        "filename": "vae/ltx-2.5-video-vae-bf16.safetensors",
        "folder": "vae"
    },
    {
        "repo_id": "Comfy-Org/gemma-4",
        "filename": "text_encoders/gemma4_e2b_it_int8_convrot.safetensors",
        "folder": "clip"
    },
    {
        "repo_id": "Lightricks/LTX-2.5",
        "filename": "text_encoders/gemma4-12b-with-proj-ltx-2.5-comfy-int8-convrot.safetensors",
        "folder": "clip"
    },
    {
        "repo_id": "Lightricks/LTX-2.5",
        "filename": "diffusion_models/ltx-2.5-22b-distilled-transformer-comfy-int8-convrot.safetensors",
        "folder": "unet"
    },
    {
        "repo_id": "Lightricks/LTX-2.5",
        "filename": "latent_upscale_models/ltx-2.5-latent-spatial-upscaler-x2-bf16-1.0.safetensors",
        "folder": "latent_upscale_models"
    }
]

try:
    for item in models:
        repo_id = item["repo_id"]
        filename = item["filename"]
        folder = item["folder"]

        dest_dir = base_models_dir / folder
        dest_dir.mkdir(parents=True, exist_ok=True)
        final_path = dest_dir / Path(filename).name

        # Jos tiedosto on symlinkki tai korruptoitunut, poistetaan se ensin puhtaan latauksen takaamiseksi
        if final_path.exists() or final_path.is_symlink():
            if final_path.is_symlink() or final_path.stat().st_size < 1000:
                print(f"Poistetaan viallinen tai tyhjä linkki: {final_path}")
                final_path.unlink(missing_ok=True)
            else:
                print(f"  Tiedosto on jo olemassa ja kunnossa: {final_path}")
                continue

        print(f"\nℹ  Ladataan: {filename} kohteeseen ComfyUI/models/{folder}/")

        # Ladataan suoraan ilman symlinkkejä local_dir-asetuksella
        downloaded = hf_hub_download(
            repo_id=repo_id,
            filename=filename,
            repo_type="model",
            token=hf_token,
            local_dir=str(dest_dir),
            local_dir_use_symlinks=False
        )

        # Jos huggingface lataa sen alihakemistorakenteella, siirretään tiedosto juureen
        downloaded_path = Path(downloaded)
        if downloaded_path.exists() and downloaded_path != final_path:
            if final_path.exists():
                final_path.unlink()
            shutil.move(str(downloaded_path), str(final_path))

        if final_path.exists() and not final_path.is_symlink():
            print(f"✅ Tallennettu onnistuneesti: {final_path}")
            print(f"Koko: {final_path.stat().st_size / (1024**3):.2f} GB")

    print("\n✅ Kaikkien tiedostojen siirto ja tarkistus valmis. Käynnistä ComfyUI uudelleen havaitaksesi uudet mallit.")

except Exception as e:
    print(f"\n❌ Virhe latauksessa: {e}")
```

```python
import os
from pathlib import Path

# Määritetään ComfyUI:n output-kansio
output_dir = Path("/content/ComfyUI/output")

if output_dir.exists():
    files = list(output_dir.glob("**/*"))
    # Suodatetaan pois pelkät kansiot ja piilotiedostot (esim. .keep-tiedostot)
    output_files = [f for f in files if f.is_file() and not f.name.startswith("_")]

    if output_files:
        print(f"Löydettiin {len(output_files)} output-tiedostoa:")
        for file in output_files:
            print(f"- {file.relative_to(output_dir.parent)}")
    else:
        print("Output-kansio on tyhjä tai sisältää vain tyhjiä mallitiedostoja.")
else:
    print("Virhe: ComfyUI:n output-kansiota ei löytynyt!")
```
